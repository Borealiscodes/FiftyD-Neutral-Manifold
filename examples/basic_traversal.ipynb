# Basic Traversal in the 50-Dimensional Manifold

This notebook demonstrates neutral coordinate traversal using the
TraversalEngine and Coordinate50D classes.

```python
from src.manifold_50d import Coordinate50D
from src.traversal_engine import TraversalEngine
from src.utils.vector_ops import add, scale
import numpy as np
```

```python
start = Coordinate50D([0.0] * 50)
start
```

```python
step_vector = Coordinate50D([0.1] * 50)
step_vector
```

```python
engine = TraversalEngine(step_size=0.1)
```

```python
coords = [start]
current = start

for _ in range(10):
    current = engine.step(current, step_vector)
    coords.append(current)

coords[-1]
```

```python
proj = np.array([[c.values[0], c.values[1]] for c in coords])
proj
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.plot(proj[:,0], proj[:,1], marker='o')
plt.title("2D Projection of 50D Traversal")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```
