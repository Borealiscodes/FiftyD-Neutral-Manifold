# Combined Demonstration

This notebook demonstrates traversal, scalar field evaluation,
stability surface classification, and operator transformations inside
the 50-dimensional manifold.

```python
from src.manifold_50d import Coordinate50D
from src.traversal_engine import TraversalEngine
from src.scalar_fields import altitude_field
from src.stability_surfaces import StabilitySurface
from src.operators import scaling_operator
import numpy as np
```

```python
# Initialize components
engine = TraversalEngine(step_size=0.1)
surface = StabilitySurface(field=altitude_field, threshold=2.0)
```

```python
# Create starting coordinate
coord = Coordinate50D([0.0] * 50)
coord
```

```python
# Define step vector
step_vector = Coordinate50D([0.05] * 50)
step_vector
```

```python
# Perform traversal steps
coords = [coord]
current = coord

for _ in range(20):
    current = engine.step(current, step_vector)
    coords.append(current)

coords[-1]
```

```python
# Evaluate scalar field
scalar_values = [altitude_field(c) for c in coords]
scalar_values[:10]
```

```python
# Classify using stability surface
classifications = [surface.classify(c) for c in coords]
classifications[:10]
```

```python
# Apply operator transformation
scale_factor = 1.2
operator_results = [scaling_operator(c, scale_factor) for c in coords]
operator_results[-1]
```

```python
# Project to 2D for visualization
proj = np.array([[c.values[0], c.values[1]] for c in coords])
proj[:5]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(proj[:,0], proj[:,1], c=scalar_values, cmap="viridis")
plt.title("Combined Demo: Scalar Field Values (2D Projection)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.colorbar(label="Scalar Value")
plt.grid(True)
plt.show()
```
