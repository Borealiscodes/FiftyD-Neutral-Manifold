# Operator Demonstration

This notebook demonstrates neutral coordinate transformations using
operator placeholders inside the 50-dimensional manifold.

```python
from src.manifold_50d import Coordinate50D
from src.operators import identity_operator, scaling_operator, shift_operator, linear_operator
import numpy as np
```

```python
# Create a sample coordinate
coord = Coordinate50D(np.random.uniform(-1.0, 1.0, 50).tolist())
coord
```

```python
# Identity operator
identity_result = identity_operator(coord)
identity_result
```

```python
# Scaling operator
scale_factor = 1.5
scaling_result = scaling_operator(coord, scale_factor)
scaling_result
```

```python
# Shift operator
shift_vector = Coordinate50D([0.05] * 50)
shift_result = shift_operator(coord, shift_vector)
shift_result
```

```python
# Linear operator using a random 50x50 matrix
matrix = np.random.uniform(-0.2, 0.2, (50, 50))
linear_result = linear_operator(coord, matrix)
linear_result
```
