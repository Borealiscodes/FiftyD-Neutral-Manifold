# Stability Surface Demonstration

This notebook demonstrates evaluating stability surfaces using scalar
fields and threshold-based classification inside the 50-dimensional
manifold.

```python
from src.manifold_50d import Coordinate50D
from src.scalar_fields import altitude_field
from src.stability_surfaces import StabilitySurface
import numpy as np
```

```python
# Create a stability surface using the altitude field
surface = StabilitySurface(field=altitude_field, threshold=2.5)
```

```python
# Generate sample coordinates
coords = [Coordinate50D(np.random.uniform(-1.0, 1.0, 50).tolist()) for _ in range(200)]
coords[:3]
```

```python
# Classify coordinates using the stability surface
classifications = [surface.classify(c) for c in coords]
classifications[:10]
```

```python
# Project coordinates to 2D for visualization
proj = np.array([[c.values[0], c.values[1]] for c in coords])
proj[:5]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(proj[:,0], proj[:,1], c=classifications, cmap="coolwarm")
plt.title("Stability Surface Classification (2D Projection)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.colorbar(label="Classification (0=Outside, 1=Inside)")
plt.show()
```
