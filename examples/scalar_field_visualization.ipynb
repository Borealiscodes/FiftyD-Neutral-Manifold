# Scalar Field Visualization in the 50-Dimensional Manifold

This notebook demonstrates evaluating scalar fields and visualizing
simple projections of scalar values.

```python
from src.manifold_50d import Coordinate50D
from src.scalar_fields import altitude_field, density_field, smoothness_field
import numpy as np
```

```python
# Generate sample coordinates
coords = [Coordinate50D(np.random.uniform(-1.0, 1.0, 50).tolist()) for _ in range(100)]
coords[:3]
```

```python
# Evaluate scalar fields
alt_values = [altitude_field(c) for c in coords]
dens_values = [density_field(c) for c in coords]
smooth_values = [smoothness_field(c) for c in coords]

alt_values[:5], dens_values[:5], smooth_values[:5]
```

```python
# Project coordinates to 2D for visualization
proj = np.array([[c.values[0], c.values[1]] for c in coords])
proj[:5]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(proj[:,0], proj[:,1], c=alt_values, cmap="viridis")
plt.title("Altitude Field (2D Projection)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.colorbar(label="Scalar Value")
plt.grid(True)
plt.show()
```

Scalar field visualization complete.
